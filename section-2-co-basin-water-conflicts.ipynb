If states are recorded together, i.e. "AZ; CA", then we risk either repeating or missing the record in our data analysis.

Individually, write down step-by-step instructions on how you would wrangle the data frame to obtain an array (without repetition) of the state codes in which a water conflict has been reported. Would you keep or remove NA values and why? It’s ok if you don’t know how to code each step - it’s way more important to have an idea of what you would like to do. :\

1. Split and explode state names, reassign to dataframe.
2. Remove space from States variables
3. Drop NAs- not giving us the info we need.
4. Filter 'Water Conflicts' to 'Y'
5. Count 'Y's per each State

```python
# 1. Filter 'Water Conflicts' to 'Y'
states_conflict = df.loc[df['Conflict Present'] == 'Y', 'State']
```

```python
# 2. Split and explode state names, reassign to dataframe.
states_split = states_conflict.str.split(';').explode()
```

```python
# 3. Count 'Y's per each State, keep NAs
states_count = states_split.value_counts()
states_count
```
